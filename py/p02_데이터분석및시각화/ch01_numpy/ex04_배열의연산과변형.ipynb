```python
# 배열의 연산과 변형
## 백터화 연산(Vectorized Operations)
import numpy as np

# 배열 생성
arr = np.arange(1, 5)
arr
```

```python
## 백터화 연산(Vectorized Operations)
arr + 2
```

```python
# 리스트
list = [1, 2, 3, 4]
list
```

```python
# 반복
list * 2
```

```python
arr = np.arange(1,5)
arr1 = np.arange(5, 9)
print(arr)
print(arr1)
```

```python
arr + arr1
```

```python
arr * arr1
```

```python
list = [1, 2, 3, 4]
list1 = [5, 6, 7, 8]
```

```python
list + list1
```

```python
list * list1
```

## 주요 산술 연산 함수
### 수학 함수들

```python
# 제곱
print(np.square(arr))  # array([ 1,  4,  9, 16])

# 제곱근(루트)
print(np.sqrt(arr))    # array([1.        , 1.41421356, 1.73205081, 2.        ])

# exp(지수승)
print(np.exp(arr))     # array([ 2.71828183,  7.3890561 , 20.08553692, 54.59815003])

# log
print(np.log(arr))     # array([0.        , 0.69314718, 1.09861229, 1.38629436])
```

## 배열 연산 함수들

```python
# 더하기
print(np.add(arr, arr1))  # arr + arr1과 동일. array([ 6,  8, 10, 12])

# 합계
print(np.sum(arr))        # 10

# 누적합
print(np.cumsum(arr))     # array([ 1,  3,  6, 10])
```

## 통계 함수들

```python
# 평균
print(np.mean(arr))     # 2.5

# 편차
print(np.var(arr))      # 1.25

# 표준편차
print(np.std(arr))      # 1.118033988749895

# 최솟값
print(np.min(arr))      # 1

# 최댓값
print(np.max(arr))      # 4

# 최솟값의 인덱스
print(np.argmin(arr))   # 0

# 최댓값의 인덱스
print(np.argmax(arr))   # 3
```

# 배열의 변형(Transformation)
## Transpose(축 변환)

```python
# 2차원 배열의 축 변환
arr = np.arange(10).reshape(2, 5)
print(arr)
```

```python
print(arr.transpose())
```

```python
print(arr.T) #arr.rtansposw()와 동일
```

```python
# 3차원 배열의 축 변환
arr3d = np.arange(27).reshape(3, 3, 3)
print(arr3d)
```

```python
# 특정 축 순서 지정
print(arr3d.transpose(1, 0, 2))
```
